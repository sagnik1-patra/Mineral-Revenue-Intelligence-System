```python
import os
import pandas as pd
import numpy as np
import json
import yaml
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder

# -------------------------------
# 📂 PATHS
# -------------------------------
DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System\mineral_wise_revenue_collected.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------------------------------
# 📊 LOAD DATA
# -------------------------------
df = pd.read_csv(DATA_PATH)
print("✅ Data Loaded:", df.shape)

# -------------------------------
# 🧹 CLEANING
# -------------------------------
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.ffill()

for col in df.columns:
    df[col] = df[col].astype(str)
    df[col] = df[col].str.replace(",", "", regex=False)
    df[col] = df[col].str.replace("₹", "", regex=False)
    df[col] = df[col].str.replace("$", "", regex=False)
    df[col] = pd.to_numeric(df[col], errors='ignore')

# -------------------------------
# 🎯 TARGET
# -------------------------------
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
if len(numeric_cols) == 0:
    raise Exception("❌ No numeric columns found")

target_col = numeric_cols[-1]
print("🎯 Target:", target_col)

# -------------------------------
# 🔤 ENCODING
# -------------------------------
for col in df.select_dtypes(include='object').columns:
    df[col] = LabelEncoder().fit_transform(df[col].astype(str))

# -------------------------------
# 🎯 FEATURES
# -------------------------------
X = df.drop(columns=[target_col])
y = df[target_col]

# -------------------------------
# 🎶 HSA FEATURE SELECTION
# -------------------------------
def hsa_feature_selection(X, y, harmony_memory_size=10, iterations=20, hmcr=0.9, par=0.3):
    n_features = X.shape[1]

    # Initialize Harmony Memory
    HM = np.random.randint(0, 2, (harmony_memory_size, n_features))
    fitness_vals = np.zeros(harmony_memory_size)

    def fitness(mask):
        if np.sum(mask) == 0:
            return -np.inf

        cols = X.columns[mask == 1]
        X_sub = X[cols]

        X_train, X_test, y_train, y_test = train_test_split(X_sub, y, test_size=0.2)

        model = RandomForestRegressor(n_estimators=100)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)

        return r2_score(y_test, preds)

    # Evaluate initial HM
    for i in range(harmony_memory_size):
        fitness_vals[i] = fitness(HM[i])

    for _ in range(iterations):
        new_harmony = np.zeros(n_features)

        for j in range(n_features):
            if np.random.rand() < hmcr:
                idx = np.random.randint(0, harmony_memory_size)
                new_harmony[j] = HM[idx][j]

                # Pitch adjustment
                if np.random.rand() < par:
                    new_harmony[j] = 1 - new_harmony[j]
            else:
                new_harmony[j] = np.random.randint(0, 2)

        new_fit = fitness(new_harmony)

        worst_idx = np.argmin(fitness_vals)

        if new_fit > fitness_vals[worst_idx]:
            HM[worst_idx] = new_harmony
            fitness_vals[worst_idx] = new_fit

    best_idx = np.argmax(fitness_vals)
    best_mask = HM[best_idx]

    return X.columns[best_mask == 1]

print("🎶 Running HSA Optimization...")
selected_features = hsa_feature_selection(X, y)

print("✅ Selected Features:", list(selected_features))

X = X[selected_features]

# -------------------------------
# 🔀 SPLIT
# -------------------------------
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# -------------------------------
# 🌳 MODEL
# -------------------------------
model = RandomForestRegressor(n_estimators=200)
model.fit(X_train, y_train)

preds = model.predict(X_test)

mae = mean_absolute_error(y_test, preds)
r2 = r2_score(y_test, preds)

print(f"📊 MAE: {mae}")
print(f"📊 R2: {r2}")

# -------------------------------
# 📄 RESULTS CSV
# -------------------------------
results_df = pd.DataFrame({
    "Metric": ["MAE", "R2"],
    "Value": [mae, r2]
})

results_df.to_csv(os.path.join(OUTPUT_DIR, "hsa_results.csv"), index=False)

# -------------------------------
# 📈 RESULTS GRAPH
# -------------------------------
plt.figure()
plt.bar(results_df["Metric"], results_df["Value"])
plt.title("HSA Model Performance")
plt.savefig(os.path.join(OUTPUT_DIR, "hsa_results_graph.png"))
plt.show()

# -------------------------------
# 📊 HEATMAP
# -------------------------------
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("HSA Heatmap")
plt.savefig(os.path.join(OUTPUT_DIR, "hsa_heatmap.png"))
plt.show()

# -------------------------------
# 📈 PREDICTIONS CSV
# -------------------------------
pred_df = pd.DataFrame({
    "actual": y_test,
    "predicted": preds
})

pred_df.to_csv(os.path.join(OUTPUT_DIR, "hsa_predictions.csv"), index=False)

# -------------------------------
# 📉 PREDICTION GRAPH
# -------------------------------
plt.figure()
plt.plot(pred_df["actual"].values, label="Actual")
plt.plot(pred_df["predicted"].values, label="Predicted")
plt.legend()
plt.title("HSA Prediction Graph")
plt.savefig(os.path.join(OUTPUT_DIR, "hsa_prediction_graph.png"))
plt.show()

# -------------------------------
# 📊 ACCURACY GRAPH
# -------------------------------
plt.figure()
plt.bar(["R2"], [r2])
plt.title("HSA Accuracy")
plt.savefig(os.path.join(OUTPUT_DIR, "hsa_accuracy_graph.png"))
plt.show()

# -------------------------------
# 💾 SAVE MODEL
# -------------------------------
joblib.dump(model, os.path.join(OUTPUT_DIR, "hsa_model.pkl"))

# -------------------------------
# 📄 YAML
# -------------------------------
config = {
    "model": "Harmony Search Optimized RandomForest",
    "features": list(selected_features),
    "target": target_col,
    "metrics": {"MAE": float(mae), "R2": float(r2)}
}

with open(os.path.join(OUTPUT_DIR, "hsa_config.yaml"), "w") as f:
    yaml.dump(config, f)

# -------------------------------
# 📦 JSON
# -------------------------------
with open(os.path.join(OUTPUT_DIR, "hsa_results.json"), "w") as f:
    json.dump(config, f, indent=4)

print("\n🚀 HSA PROJECT COMPLETE! All files saved with 'hsa_' prefix")
```
