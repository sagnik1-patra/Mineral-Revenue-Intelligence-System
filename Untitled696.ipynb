```python
import os
import pandas as pd
import numpy as np
import json
import yaml
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder

# -------------------------------
# 📂 PATHS
# -------------------------------
DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System\mineral_wise_revenue_collected.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------------------------------
# 📊 LOAD DATA
# -------------------------------
df = pd.read_csv(DATA_PATH)
print("✅ Data Loaded:", df.shape)

# -------------------------------
# 🧹 CLEANING
# -------------------------------
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.ffill()

for col in df.columns:
    df[col] = df[col].astype(str)
    df[col] = df[col].str.replace(",", "", regex=False)
    df[col] = df[col].str.replace("₹", "", regex=False)
    df[col] = df[col].str.replace("$", "", regex=False)
    df[col] = pd.to_numeric(df[col], errors='ignore')

# -------------------------------
# 🎯 TARGET
# -------------------------------
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
if len(numeric_cols) == 0:
    raise Exception("❌ No numeric columns found")

target_col = numeric_cols[-1]
print("🎯 Target:", target_col)

# -------------------------------
# 🔤 ENCODING
# -------------------------------
for col in df.select_dtypes(include='object').columns:
    df[col] = LabelEncoder().fit_transform(df[col].astype(str))

# -------------------------------
# 🎯 FEATURES
# -------------------------------
X = df.drop(columns=[target_col])
y = df[target_col]

# -------------------------------
# 🐺 GWO FEATURE SELECTION
# -------------------------------
def gwo_feature_selection(X, y, n_wolves=10, iterations=20):
    n_features = X.shape[1]

    wolves = np.random.rand(n_wolves, n_features)
    fitness_scores = np.zeros(n_wolves)

    def fitness(position):
        mask = (position > 0.5).astype(int)
        if np.sum(mask) == 0:
            return -np.inf

        cols = X.columns[mask == 1]
        X_sub = X[cols]

        X_train, X_test, y_train, y_test = train_test_split(X_sub, y, test_size=0.2)

        model = RandomForestRegressor(n_estimators=100)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)

        return r2_score(y_test, preds)

    # Initial fitness
    for i in range(n_wolves):
        fitness_scores[i] = fitness(wolves[i])

    # Identify alpha, beta, delta
    def get_leaders(wolves, scores):
        sorted_idx = np.argsort(scores)[::-1]
        return wolves[sorted_idx[0]], wolves[sorted_idx[1]], wolves[sorted_idx[2]]

    alpha, beta, delta = get_leaders(wolves, fitness_scores)

    # Main loop
    for t in range(iterations):
        a = 2 - t * (2 / iterations)

        for i in range(n_wolves):
            for j in range(n_features):
                r1, r2 = np.random.rand(), np.random.rand()
                A1 = 2 * a * r1 - a
                C1 = 2 * r2

                D_alpha = abs(C1 * alpha[j] - wolves[i][j])
                X1 = alpha[j] - A1 * D_alpha

                r1, r2 = np.random.rand(), np.random.rand()
                A2 = 2 * a * r1 - a
                C2 = 2 * r2

                D_beta = abs(C2 * beta[j] - wolves[i][j])
                X2 = beta[j] - A2 * D_beta

                r1, r2 = np.random.rand(), np.random.rand()
                A3 = 2 * a * r1 - a
                C3 = 2 * r2

                D_delta = abs(C3 * delta[j] - wolves[i][j])
                X3 = delta[j] - A3 * D_delta

                wolves[i][j] = (X1 + X2 + X3) / 3

            wolves[i] = np.clip(wolves[i], 0, 1)

        # Re-evaluate
        for i in range(n_wolves):
            fitness_scores[i] = fitness(wolves[i])

        alpha, beta, delta = get_leaders(wolves, fitness_scores)

    best_mask = (alpha > 0.5).astype(int)
    return X.columns[best_mask == 1]

print("🐺 Running GWO Optimization...")
selected_features = gwo_feature_selection(X, y)

print("✅ Selected Features:", list(selected_features))

X = X[selected_features]

# -------------------------------
# 🔀 SPLIT
# -------------------------------
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# -------------------------------
# 🌳 MODEL
# -------------------------------
model = RandomForestRegressor(n_estimators=200)
model.fit(X_train, y_train)

preds = model.predict(X_test)

mae = mean_absolute_error(y_test, preds)
r2 = r2_score(y_test, preds)

print(f"📊 MAE: {mae}")
print(f"📊 R2: {r2}")

# -------------------------------
# 📄 RESULTS CSV
# -------------------------------
results_df = pd.DataFrame({
    "Metric": ["MAE", "R2"],
    "Value": [mae, r2]
})

results_df.to_csv(os.path.join(OUTPUT_DIR, "gwoa_results.csv"), index=False)

# -------------------------------
# 📈 RESULTS GRAPH
# -------------------------------
plt.figure()
plt.bar(results_df["Metric"], results_df["Value"])
plt.title("GWOA Model Performance")
plt.savefig(os.path.join(OUTPUT_DIR, "gwoa_results_graph.png"))
plt.show()

# -------------------------------
# 📊 HEATMAP
# -------------------------------
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("GWOA Heatmap")
plt.savefig(os.path.join(OUTPUT_DIR, "gwoa_heatmap.png"))
plt.show()

# -------------------------------
# 📈 PREDICTIONS CSV
# -------------------------------
pred_df = pd.DataFrame({
    "actual": y_test,
    "predicted": preds
})

pred_df.to_csv(os.path.join(OUTPUT_DIR, "gwoa_predictions.csv"), index=False)

# -------------------------------
# 📉 PREDICTION GRAPH
# -------------------------------
plt.figure()
plt.plot(pred_df["actual"].values, label="Actual")
plt.plot(pred_df["predicted"].values, label="Predicted")
plt.legend()
plt.title("GWOA Prediction Graph")
plt.savefig(os.path.join(OUTPUT_DIR, "gwoa_prediction_graph.png"))
plt.show()

# -------------------------------
# 📊 ACCURACY GRAPH
# -------------------------------
plt.figure()
plt.bar(["R2"], [r2])
plt.title("GWOA Accuracy")
plt.savefig(os.path.join(OUTPUT_DIR, "gwoa_accuracy_graph.png"))
plt.show()

# -------------------------------
# 💾 SAVE MODEL
# -------------------------------
joblib.dump(model, os.path.join(OUTPUT_DIR, "gwoa_model.pkl"))

# -------------------------------
# 📄 YAML
# -------------------------------
config = {
    "model": "GWOA Optimized RandomForest",
    "features": list(selected_features),
    "target": target_col,
    "metrics": {"MAE": float(mae), "R2": float(r2)}
}

with open(os.path.join(OUTPUT_DIR, "gwoa_config.yaml"), "w") as f:
    yaml.dump(config, f)

# -------------------------------
# 📦 JSON
# -------------------------------
with open(os.path.join(OUTPUT_DIR, "gwoa_results.json"), "w") as f:
    json.dump(config, f, indent=4)

print("\n🚀 GWOA PROJECT COMPLETE! All files saved with 'gwoa_' prefix")
```
