```python
import os
import pandas as pd
import numpy as np
import json
import yaml
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder

# -------------------------------
# 📂 PATHS
# -------------------------------
DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System\mineral_wise_revenue_collected.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------------------------------
# 📊 LOAD DATA
# -------------------------------
df = pd.read_csv(DATA_PATH)
print("✅ Data Loaded:", df.shape)

# -------------------------------
# 🧹 CLEANING
# -------------------------------
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.ffill()

# Convert strings → numeric
for col in df.columns:
    df[col] = df[col].astype(str)
    df[col] = df[col].str.replace(",", "", regex=False)
    df[col] = df[col].str.replace("₹", "", regex=False)
    df[col] = df[col].str.replace("$", "", regex=False)
    df[col] = pd.to_numeric(df[col], errors='ignore')

# -------------------------------
# 🎯 TARGET
# -------------------------------
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
if len(numeric_cols) == 0:
    raise Exception("❌ No numeric columns found")

target_col = numeric_cols[-1]
print("🎯 Target:", target_col)

# -------------------------------
# 🔤 ENCODING
# -------------------------------
for col in df.select_dtypes(include='object').columns:
    df[col] = LabelEncoder().fit_transform(df[col].astype(str))

# -------------------------------
# 🎯 FEATURES
# -------------------------------
X = df.drop(columns=[target_col])
y = df[target_col]

# -------------------------------
# 🐦 CSA FEATURE SELECTION
# -------------------------------
def csa_feature_selection(X, y, n_crows=10, iterations=20, flight_length=2, awareness_prob=0.1):
    n_features = X.shape[1]

    # Initialize population
    positions = np.random.randint(0, 2, (n_crows, n_features))
    memory = positions.copy()
    fitness = np.zeros(n_crows)

    def evaluate(mask):
        if np.sum(mask) == 0:
            return -np.inf
        cols = X.columns[mask == 1]
        X_sub = X[cols]

        X_train, X_test, y_train, y_test = train_test_split(X_sub, y, test_size=0.2)

        model = RandomForestRegressor(n_estimators=100)
        model.fit(X_train, y_train)
        preds = model.predict(X_test)

        return r2_score(y_test, preds)

    # Initial fitness
    for i in range(n_crows):
        fitness[i] = evaluate(positions[i])

    for _ in range(iterations):
        for i in range(n_crows):
            j = np.random.randint(0, n_crows)

            if np.random.rand() > awareness_prob:
                new_pos = positions[i] + flight_length * (memory[j] - positions[i])
                new_pos = np.where(new_pos > 0.5, 1, 0)
            else:
                new_pos = np.random.randint(0, 2, n_features)

            new_fit = evaluate(new_pos)

            if new_fit > fitness[i]:
                positions[i] = new_pos
                fitness[i] = new_fit
                memory[i] = new_pos

    best_idx = np.argmax(fitness)
    best_mask = memory[best_idx]

    return X.columns[best_mask == 1]

print("🐦 Running CSA Optimization...")
selected_features = csa_feature_selection(X, y)

print("✅ Selected Features:", list(selected_features))

X = X[selected_features]

# -------------------------------
# 🔀 SPLIT
# -------------------------------
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# -------------------------------
# 🌳 MODEL
# -------------------------------
model = RandomForestRegressor(n_estimators=200)
model.fit(X_train, y_train)

preds = model.predict(X_test)

mae = mean_absolute_error(y_test, preds)
r2 = r2_score(y_test, preds)

print(f"📊 MAE: {mae}")
print(f"📊 R2: {r2}")

# -------------------------------
# 📄 RESULTS CSV
# -------------------------------
results_df = pd.DataFrame({
    "Metric": ["MAE", "R2"],
    "Value": [mae, r2]
})

results_df.to_csv(os.path.join(OUTPUT_DIR, "csa_results.csv"), index=False)

# -------------------------------
# 📈 RESULTS GRAPH
# -------------------------------
plt.figure()
plt.bar(results_df["Metric"], results_df["Value"])
plt.title("CSA Model Performance")
plt.savefig(os.path.join(OUTPUT_DIR, "csa_results_graph.png"))
plt.show()

# -------------------------------
# 📊 HEATMAP
# -------------------------------
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), cmap="coolwarm")
plt.title("CSA Heatmap")
plt.savefig(os.path.join(OUTPUT_DIR, "csa_heatmap.png"))
plt.show()

# -------------------------------
# 📈 PREDICTIONS CSV
# -------------------------------
pred_df = pd.DataFrame({
    "actual": y_test,
    "predicted": preds
})

pred_df.to_csv(os.path.join(OUTPUT_DIR, "csa_predictions.csv"), index=False)

# -------------------------------
# 📉 PREDICTION GRAPH
# -------------------------------
plt.figure()
plt.plot(pred_df["actual"].values, label="Actual")
plt.plot(pred_df["predicted"].values, label="Predicted")
plt.legend()
plt.title("CSA Prediction Graph")
plt.savefig(os.path.join(OUTPUT_DIR, "csa_prediction_graph.png"))
plt.show()

# -------------------------------
# 📊 ACCURACY GRAPH
# -------------------------------
plt.figure()
plt.bar(["R2"], [r2])
plt.title("CSA Accuracy")
plt.savefig(os.path.join(OUTPUT_DIR, "csa_accuracy_graph.png"))
plt.show()

# -------------------------------
# 💾 SAVE MODEL
# -------------------------------
joblib.dump(model, os.path.join(OUTPUT_DIR, "csa_model.pkl"))

# -------------------------------
# 📄 YAML
# -------------------------------
config = {
    "model": "CSA-Optimized RandomForest",
    "features": list(selected_features),
    "target": target_col,
    "metrics": {"MAE": float(mae), "R2": float(r2)}
}

with open(os.path.join(OUTPUT_DIR, "csa_config.yaml"), "w") as f:
    yaml.dump(config, f)

# -------------------------------
# 📦 JSON
# -------------------------------
with open(os.path.join(OUTPUT_DIR, "csa_results.json"), "w") as f:
    json.dump(config, f, indent=4)

print("\n🚀 CSA PROJECT COMPLETE! All files saved with 'csa_' prefix")
```
