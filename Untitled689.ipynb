```python
import os
import pandas as pd
import numpy as np
import json
import yaml
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.preprocessing import LabelEncoder

# -------------------------------
# 📂 PATHS
# -------------------------------
DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System\mineral_wise_revenue_collected.csv"
OUTPUT_DIR = r"C:\Users\NXTWAVE\Downloads\Mineral Revenue Intelligence System"

os.makedirs(OUTPUT_DIR, exist_ok=True)

# -------------------------------
# 📊 LOAD DATA
# -------------------------------
df = pd.read_csv(DATA_PATH)
print("✅ Data Loaded:", df.shape)

# -------------------------------
# 🧹 CLEANING
# -------------------------------
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.ffill()

# Convert to numeric
for col in df.columns:
    df[col] = df[col].astype(str)
    df[col] = df[col].str.replace(",", "", regex=False)
    df[col] = df[col].str.replace("₹", "", regex=False)
    df[col] = df[col].str.replace("$", "", regex=False)
    df[col] = pd.to_numeric(df[col], errors='ignore')

# -------------------------------
# 🎯 TARGET SELECTION
# -------------------------------
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

if len(numeric_cols) == 0:
    raise Exception("❌ No numeric columns found")

target_col = numeric_cols[-1]
print("🎯 Target:", target_col)

# -------------------------------
# 🔤 ENCODING
# -------------------------------
for col in df.select_dtypes(include='object').columns:
    df[col] = LabelEncoder().fit_transform(df[col].astype(str))

# -------------------------------
# 🎯 FEATURES
# -------------------------------
X = df.drop(columns=[target_col])
y = df[target_col]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# -------------------------------
# 🤖 MODELS
# -------------------------------
models = {
    "RandomForest": RandomForestRegressor(n_estimators=200),
    "LinearRegression": LinearRegression()
}

results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    mae = mean_absolute_error(y_test, preds)
    r2 = r2_score(y_test, preds)

    results[name] = {
        "model": model,
        "mae": mae,
        "r2": r2,
        "preds": preds
    }

# -------------------------------
# 📊 RESULTS CSV
# -------------------------------
results_df = pd.DataFrame({
    name: [results[name]["mae"], results[name]["r2"]]
    for name in results
}, index=["MAE", "R2"])

results_path = os.path.join(OUTPUT_DIR, "model_results.csv")
results_df.to_csv(results_path)

# -------------------------------
# 📈 COMPARISON GRAPH
# -------------------------------
results_df.T.plot(kind="bar")
plt.title("Model Comparison")
plt.savefig(os.path.join(OUTPUT_DIR, "model_comparison.png"))
plt.show()

# -------------------------------
# 📊 HEATMAP
# -------------------------------
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(), annot=True, cmap="coolwarm")
plt.title("Feature Correlation Heatmap")
plt.savefig(os.path.join(OUTPUT_DIR, "heatmap.png"))
plt.show()

# -------------------------------
# 📈 PREDICTIONS (BEST MODEL)
# -------------------------------
best_model_name = max(results, key=lambda x: results[x]["r2"])
best_model = results[best_model_name]["model"]
best_preds = results[best_model_name]["preds"]

pred_df = pd.DataFrame({
    "actual": y_test.values,
    "predicted": best_preds
})

pred_path = os.path.join(OUTPUT_DIR, "predictions.csv")
pred_df.to_csv(pred_path, index=False)

# -------------------------------
# 📉 PREDICTION GRAPH
# -------------------------------
plt.figure()
plt.plot(pred_df["actual"].values, label="Actual")
plt.plot(pred_df["predicted"].values, label="Predicted")
plt.legend()
plt.title("Prediction vs Actual")
plt.savefig(os.path.join(OUTPUT_DIR, "prediction_graph.png"))
plt.show()

# -------------------------------
# 📊 ACCURACY GRAPH
# -------------------------------
r2_scores = [results[m]["r2"] for m in results]

plt.figure()
plt.bar(results.keys(), r2_scores)
plt.title("Accuracy (R2 Score)")
plt.savefig(os.path.join(OUTPUT_DIR, "accuracy_graph.png"))
plt.show()

# -------------------------------
# 📈 RESULT GRAPH (MAE)
# -------------------------------
mae_scores = [results[m]["mae"] for m in results]

plt.figure()
plt.bar(results.keys(), mae_scores)
plt.title("Error (MAE)")
plt.savefig(os.path.join(OUTPUT_DIR, "error_graph.png"))
plt.show()

# -------------------------------
# 💾 SAVE BEST MODEL
# -------------------------------
joblib.dump(best_model, os.path.join(OUTPUT_DIR, "best_model.pkl"))

# -------------------------------
# 📄 YAML
# -------------------------------
config = {
    "best_model": best_model_name,
    "target": target_col,
    "metrics": results_df.to_dict()
}

with open(os.path.join(OUTPUT_DIR, "config.yaml"), "w") as f:
    yaml.dump(config, f)

# -------------------------------
# 📦 JSON
# -------------------------------
with open(os.path.join(OUTPUT_DIR, "results.json"), "w") as f:
    json.dump(results_df.to_dict(), f, indent=4)

print("\n🚀 ALL VISUALS + FILES GENERATED!")
```
